## loop

# Functions

```python
def square(size):
    for i in range(4):
        forward(size)
        right(90)
        print("i is ",i)
```

```python
reset()
print("here")
square(80)
print("there")
```

```python
reset()
print("here")
square(20)
forward(60)
square(20)
print("there")
```

```python
reset()
print("here")
square(20)

penup()
forward(60)
pendown()

square(20)
print("there")
```

```python
def penup_forward(length):
    penup()
    forward(length)
    pendown()
```

```python
reset()
square(20)

penup_forward(60)

square(20)
```
